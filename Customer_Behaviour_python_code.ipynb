```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("customer_shopping_behavior.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
```

```python
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
lable=['young_Adult','Adult','Mid_Age','Old']
df['Age_Group']=pd.qcut(df['age'],4,labels=lable)
df['Age_Group']
```

```python
df['frequency_of_purchases'].unique()
frequency_mapping={
    'Weekly':7,
    'Fortnightly':14,
    "Annually":365,
    'Quarterly':90,
    "Bi-Weekly":14,
    'Monthly':30,
    'Every 3 Months':90
}
df['purchase_freq']=df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_freq','frequency_of_purchases']]
```

```python
# df[['discount_applied','promo_code_used'
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df.drop('promo_code_used',axis=1,inplace=True)
```
